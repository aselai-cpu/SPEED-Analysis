## 3. Understand Codebook Structure and Create Mapping Dictionary

```python
# Function to create enum mapping from codebook
def create_enum_mapping(codebook_sheets):
    """
    Create a dictionary mapping column names to their descriptions and enum values.
    
    Structure:
    {
        column_name: {
            "description": "...",
            "enums": {
                "key": "value",
                ...
            }
        }
    }
    
    Reads from:
    - "Variable Description" sheet: Variable -> Variable Description
    - "Value Labels" sheet: Variable (when not NaN) indicates variable name,
                            Variable Labels contains enum definitions like '0="Missing"'
    """
    import re
    
    enum_mappings = {}
    
    # Step 1: Read Variable Description sheet to get descriptions
    var_desc_sheet = codebook_sheets.get('Variable Description')
    descriptions = {}
    
    if var_desc_sheet is not None:
        # Find the column names
        var_col = None
        desc_col = None
        for col in var_desc_sheet.columns:
            col_lower = str(col).lower()
            if 'variable' in col_lower and 'description' not in col_lower:
                var_col = col
            elif 'description' in col_lower:
                desc_col = col
        
        if var_col and desc_col:
            for _, row in var_desc_sheet.iterrows():
                var_name = row[var_col]
                desc = row[desc_col]
                if pd.notna(var_name) and pd.notna(desc):
                    var_name = str(var_name).strip().lower()
                    descriptions[var_name] = str(desc).strip()
    
    # Step 2: Read Value Labels sheet to get enum mappings
    value_labels_sheet = codebook_sheets.get('Value Labels')
    
    if value_labels_sheet is not None:
        # Find the column names
        var_col = None
        labels_col = None
        for col in value_labels_sheet.columns:
            col_lower = str(col).lower()
            if 'variable' in col_lower and 'label' not in col_lower:
                var_col = col
            elif 'label' in col_lower:
                labels_col = col
        
        if var_col and labels_col:
            current_variable = None
            
            for _, row in value_labels_sheet.iterrows():
                var_name = row[var_col]
                label_text = row[labels_col]
                
                # If Variable column has a value, it's a new variable
                if pd.notna(var_name):
                    current_variable = str(var_name).strip().lower()
                    # Initialize if not exists
                    if current_variable not in enum_mappings:
                        enum_mappings[current_variable] = {
                            "description": descriptions.get(current_variable, ""),
                            "enums": {}
                        }
                
                # Parse the Variable Labels column for enum definitions
                # Format: '0="Missing"', '1="Political Expression"', '2=\'One Day\'', '3= NULL'
                if pd.notna(label_text) and current_variable:
                    label_text = str(label_text).strip()
                    
                    # Pattern to match: key="value" or key='value' or key= NULL (case-insensitive) or key=value
                    # Match patterns like: 0="Missing", 1='One Day', 3= NULL, 4=NULL
                    # More flexible pattern to handle various formats
                    pattern = r'^(\d+|\w+)\s*=\s*(?:"([^"]*)"|\'([^\']*)\'|(NULL|null)\s*|(.+?)(?:\s*$))'
                    match = re.match(pattern, label_text, re.IGNORECASE)
                    
                    if match:
                        enum_key = match.group(1).strip()
                        # Get the value from whichever group matched (group 2=double quotes, 3=single quotes, 4=NULL, 5=unquoted)
                        enum_value = (match.group(2) or match.group(3) or 
                                     (match.group(4) if match.group(4) else "") or
                                     (match.group(5).strip() if match.group(5) else ""))
                        
                        if enum_key:
                            # Normalize NULL values
                            if enum_value.upper() == "NULL" or enum_value == "":
                                enum_value = "NULL"
                            enum_mappings[current_variable]["enums"][enum_key] = enum_value.strip()
    
    # Step 3: Also add variables that have descriptions but no enum values
    for var_name, desc in descriptions.items():
        if var_name not in enum_mappings:
            enum_mappings[var_name] = {
                "description": desc,
                "enums": {}
            }
    
    return enum_mappings

# Create the mapping
enum_mappings = create_enum_mapping(codebook_sheets)
print(f"\n\nFound {len(enum_mappings)} variables with mappings")
print(f"\nSample mappings:")
for var_name, mapping in list(enum_mappings.items())[:]:
    print(f"\n{var_name}:")
    print(f"  Description: {mapping['description'][:80]}..." if len(mapping['description']) > 80 else f"  Description: {mapping['description']}")
    print(f"  Enums: {len(mapping['enums'])} values")
    if mapping['enums']:
        sample_enums = dict(list(mapping['enums'].items())[:3])
        print(f"    Sample: {sample_enums}")
```

## 4. Take Sample Rows and Resolve Enum Values

```python
# Function to resolve enum values in a dataframe
def resolve_enums(df, enum_mappings, sample_size=5):
    """
    Take sample rows and create a new dataframe with resolved enum values.
    
    enum_mappings structure:
    {
        column_name: {
            "description": "...",
            "enums": {key: value}
        }
    }
    """
    # Take sample rows
    sample_df = df.head(sample_size).copy()
    
    # Create a resolved version
    resolved_df = sample_df.copy()
    
    # For each column in the sample, try to resolve if mapping exists
    resolved_info = {}
    
    for col in sample_df.columns:
        # Try to find matching enum mapping (case-insensitive)
        matching_var = None
        for var_name in enum_mappings.keys():
            if col.lower() == var_name.lower() or col.lower().replace('_', '') == var_name.lower().replace('_', ''):
                matching_var = var_name
                break
        
        if matching_var and enum_mappings[matching_var].get("enums"):
            # Get the enum mappings for this variable
            enum_dict = enum_mappings[matching_var]["enums"]
            
            # Create resolved column
            resolved_col_name = f"{col}_resolved"
            
            def resolve_enum_value(value):
                """Resolve enum value, handling float to int conversion"""
                if pd.isna(value):
                    return value
                
                # Convert to string and strip whitespace
                value_str = str(value).strip()
                
                # Try original value as string first (handles cases like "1.0" as string)
                if value_str in enum_dict:
                    return enum_dict[value_str]
                
                # Try converting to numeric and then to int for enum lookup
                try:
                    # First try as float
                    float_val = float(value_str)
                    
                    # If it's a whole number, try as int string
                    if float_val.is_integer():
                        int_val = int(float_val)
                        int_str = str(int_val)
                        if int_str in enum_dict:
                            return enum_dict[int_str]
                    
                    # Also try the float string representation (e.g., "22.0")
                    float_str = str(int(float_val)) if float_val.is_integer() else value_str
                    if float_str in enum_dict:
                        return enum_dict[float_str]
                        
                except (ValueError, OverflowError, TypeError):
                    pass
                
                # If value is already a float/numeric type, try direct int conversion
                try:
                    if isinstance(value, (float, int)) and not pd.isna(value):
                        int_val = int(value)
                        int_str = str(int_val)
                        if int_str in enum_dict:
                            return enum_dict[int_str]
                except (ValueError, OverflowError):
                    pass
                
                # Return original value if no match found (could be a value not in codebook)
                return value
            
            resolved_df[resolved_col_name] = sample_df[col].apply(resolve_enum_value)
            resolved_info[col] = {
                'mapping_var': matching_var,
                'description': enum_mappings[matching_var].get("description", ""),
                'unique_values': sample_df[col].unique(),
                'resolved_values': resolved_df[resolved_col_name].unique()
            }
    
    return resolved_df, resolved_info

# Get sample rows with resolved enums
sample_size = 62141
resolved_df, resolved_info = resolve_enums(df, enum_mappings, sample_size=sample_size)

print(f"Sample of {sample_size} rows with resolved enum values:\n")
display(resolved_df)
```

## 4.1. Diagnostic: Check Resolution Status

```python
# Diagnostic: Check which values were resolved and which weren't
print("Resolution Diagnostic:\n")
print("=" * 80)

for col in resolved_info.keys():
    print(f"\nColumn: {col}")
    print(f"  Mapping Variable: {resolved_info[col]['mapping_var']}")
    
    # Check original vs resolved values
    original_vals = resolved_df[col].dropna().unique()
    resolved_vals = resolved_df[f"{col}_resolved"].dropna().unique()
    
    # Find values that weren't resolved (original value == resolved value and it's numeric)
    unresolved = []
    for idx in resolved_df.index:
        orig = resolved_df.loc[idx, col]
        resolved = resolved_df.loc[idx, f"{col}_resolved"]
        if pd.notna(orig) and pd.notna(resolved):
            # If resolved value is the same as original (numeric), it wasn't resolved
            try:
                if float(orig) == float(resolved):
                    unresolved.append(orig)
            except (ValueError, TypeError):
                if str(orig) == str(resolved):
                    unresolved.append(orig)
    
    if unresolved:
        print(f"  ⚠ Unresolved values: {set(unresolved)}")
        # Check if these values exist in enum_dict
        enum_dict = enum_mappings[resolved_info[col]['mapping_var']].get("enums", {})
        for val in set(unresolved):
            val_str = str(int(val)) if isinstance(val, float) and val.is_integer() else str(val)
            if val_str not in enum_dict:
                print(f"    - {val} ({val_str}) not found in enum mappings")
            else:
                print(f"    - {val} ({val_str}) exists in mappings but wasn't resolved")
    else:
        print(f"  ✓ All non-null values resolved")
    
    print(f"  Original unique values: {len(original_vals)}")
    print(f"  Resolved unique values: {len(resolved_vals)}")
    print("-" * 80)
```

## 8.1. Export with Replaced Values (Updated)

```python
from datetime import datetime
import os

# Create output directory if it doesn't exist
output_dir = Path('output')
output_dir.mkdir(exist_ok=True)

# Generate filename with datetimestamp
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
output_filename = output_dir / f"resolved_data_{timestamp}.csv"

# Write resolved dataframe to CSV
# Replace original columns with resolved values (instead of having both)
if 'resolved_df' in locals() and not resolved_df.empty:
    # Create a copy of the resolved dataframe
    export_df = resolved_df.copy()
    
    # Replace original columns with resolved values
    for col in resolved_info.keys():
        resolved_col = f"{col}_resolved"
        if resolved_col in export_df.columns:
            # Replace the original column with resolved values
            export_df[col] = export_df[resolved_col]
            # Drop the _resolved column
            export_df = export_df.drop(columns=[resolved_col])
    
    # Rename columns to their descriptions from codebook
    column_rename_map = {}
    for col in export_df.columns:
        # Try to find matching variable in enum_mappings
        matching_var = None
        for var_name in enum_mappings.keys():
            if col.lower() == var_name.lower() or col.lower().replace('_', '') == var_name.lower().replace('_', ''):
                matching_var = var_name
                break
        
        if matching_var and enum_mappings[matching_var].get('description'):
            # Use the description as the new column name
            description = enum_mappings[matching_var]['description']
            # Clean up description for CSV column name (remove special chars, limit length)
            # Keep it readable but make it a valid column name
            clean_description = description.replace('\n', ' ').replace('\r', ' ').strip()
            # If description is too long, truncate it
            if len(clean_description) > 100:
                clean_description = clean_description[:97] + "..."
            column_rename_map[col] = clean_description
    
    # Apply the renaming
    if column_rename_map:
        export_df = export_df.rename(columns=column_rename_map)
    
    # Write to CSV
    export_df.to_csv(output_filename, index=False)
    print(f"✓ Resolved data exported successfully!")
    print(f"  File: {output_filename}")
    print(f"  Rows: {len(export_df):,}")
    print(f"  Columns: {len(export_df.columns)}")
    print(f"  Resolved columns: {len(resolved_info)}")
    print(f"  Note: Original enum values have been replaced with resolved text values")
    print(f"  Note: Column names have been replaced with their descriptions from codebook")
    
    # Also create a summary file
    summary_filename = output_dir / f"resolution_summary_{timestamp}.txt"
    with open(summary_filename, 'w') as f:
        f.write(f"SPEED Dataset - Enum Resolution Summary\n")
        f.write(f"{'='*80}\n\n")
        f.write(f"Generated: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
        f.write(f"Source data: {data_file}\n")
        f.write(f"Codebook: {codebook_file}\n\n")
        f.write(f"Dataset Summary:\n")
        f.write(f"  Total rows in dataset: {len(df):,}\n")
        f.write(f"  Sample rows exported: {len(export_df):,}\n")
        f.write(f"  Total columns: {len(df.columns)}\n")
        f.write(f"  Columns with enum mappings: {len(resolved_info)}\n")
        f.write(f"  Note: Original enum values have been replaced with resolved text values\n")
        f.write(f"  Note: Column names have been replaced with their descriptions from codebook\n\n")
        f.write(f"Resolved Columns:\n")
        f.write(f"{'-'*80}\n")
        for col, info in resolved_info.items():
            f.write(f"\n{col}:\n")
            f.write(f"  Mapping Variable: {info['mapping_var']}\n")
            if info.get('description'):
                f.write(f"  Description: {info['description']}\n")
            f.write(f"  Unique values in sample: {len(info['unique_values'])}\n")
            if enum_mappings[info['mapping_var']].get('enums'):
                enum_count = len(enum_mappings[info['mapping_var']]['enums'])
                f.write(f"  Total enum values in codebook: {enum_count}\n")
    
    print(f"✓ Summary file created: {summary_filename}")
else:
    print("⚠ No resolved data available to export. Run the previous cells first.")
```
